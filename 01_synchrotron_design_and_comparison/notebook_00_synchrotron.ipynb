# Synchrotron design and comparison

Design and simulate two **1.7 GeV electron storage rings** with the same circumference (240 m), number of cells (16), and number of dipoles (32).

Compare two cell designs: **FODO** and **double-bend achromat (DBA)**.

Work through the notebooks in this order:

1. [Build a FODO ring](notebook_01_design_fodo.ipynb): assemble the lattice, match the cell phase advances, and correct the chromaticities.
2. [Build a DBA ring](notebook_02_design_dba.ipynb): match the cell phase advances and zero dispersion in the straight, then correct the chromaticities.
3. [Radiation equilibrium and lattice comparison](notebook_03_radiation_integrals.ipynb): compare emittances, energy loss, momentum spread, and damping times using radiation integrals.
4. [Dynamic aperture](notebook_04_dynamic_aperture.ipynb): track on-momentum particles and compare survival, one ring at a time. Use each ring's horizontal equilibrium emittance as the reference for both planes.

**Optional exercises:**

5. [RF and quantum radiation](notebook_05_optional_rf_and_quantum_radiation.ipynb): add an RF cavity to the DBA ring and observe damping and quantum excitation.
6. [Misalignments and vertical emittance](notebook_06_optional_misalignments_and_vertical_emittance.ipynb): apply random quadrupole shifts to the DBA ring and inspect the resulting vertical emittance, orbit, and dispersion.
