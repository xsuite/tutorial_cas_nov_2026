# FCC-ee optics and beam–beam interaction

Compare two optics for the same FCC-ee Z lattice: the nominal optics and optics with eight times larger beta functions at the interaction points (labelled “Larger IP beta”). The tunes and corrected linear chromaticities have been restored to the same values.

Then use the nominal optics to study the effect of beam–beam interactions and crab sextupoles on particle survival.

All exercises use the same prepared lattice with simplified magnet models for faster tracking.

Use the notebooks in this folder:

1. [Optics comparison](notebook_01_optics_comparison.ipynb): inspect the optics and compare corrected and natural chromaticities.
2. [Sensitivity to alignment errors](notebook_02_alignment_sensitivity.ipynb): apply a vertical shift to one final-doublet quadrupole and compare the vertical closed orbits.
3. [Beam–beam interaction and dynamic aperture](notebook_03_beambeam_dynamic_aperture.ipynb): compare particle survival with beam–beam interactions and crab sextupoles switched on and off.
4. [Optional: increase the IP beta functions](notebook_04_optional_larger_ip_beta.ipynb): build optics with five times larger IP beta using open Twiss and matching.
